# AI/Tech Articles: loading the dataset

The quickest way is HuggingFace `datasets`; if you have this repo checked out,
`ai_tech_articles.load_dataframe()` reads the local parquet instead (which may be
newer than what is published).


```python
# Option A: straight from HuggingFace (pip install datasets)
# from datasets import load_dataset
# df = load_dataset("siavava/ai-tech-articles", split="train").to_pandas()

# Option B: the local parquet when the repo is checked out, HuggingFace otherwise
from ai_tech_articles import load_dataframe

df = load_dataframe()
print(f"{len(df):,} articles, columns: {list(df.columns)}")
df.head(10)
```

Each row is one article. `id` is the row number (sequential, collection order) and `url` is the
stable key across releases. `year` is always set; `date` (YYYY-MM-DD) is set for everything
collected since 2023 and null for the original 2023 collection.


```python
df[["id", "year", "date", "title"]].sample(8, random_state=0)
```

```python
# Text of one article
row = df[df.date.notna()].iloc[0]
print(row.title, "|", row.url, "\n")
print(row.text[:1200])
```
